```python
# ============================================================
# FITNESSML MASTER THESIS
# VERSION CONTROL (99_git.ipynb)
# ============================================================

from google.colab import drive
from google.colab import userdata

from pathlib import Path
import subprocess
import os
from datetime import datetime

# ============================================================
# SETTINGS
# ============================================================

PROJECT_DIR = Path("/content/drive/MyDrive/FitnessML_Master")

USERNAME = "HedgeHog777"
EMAIL = "olesnabok@gmail.com"
REPOSITORY = "ai-diploma"

# Якщо True → створить резервну гілку
MAKE_BACKUP_BRANCH = False

# ============================================================
# MOUNT GOOGLE DRIVE
# ============================================================

drive.mount("/content/drive", force_remount=False)

os.chdir(PROJECT_DIR)

print(f"✓ Project directory: {PROJECT_DIR}")

# ============================================================
# LOAD GITHUB TOKEN
# ============================================================

TOKEN = userdata.get("GH_TOKEN")

if not TOKEN:
    raise ValueError(
        "GitHub token not found.\n"
        "Add GH_TOKEN in Colab Secrets."
    )

REMOTE_URL = (
    f"https://{USERNAME}:{TOKEN}"
    f"@github.com/{USERNAME}/{REPOSITORY}.git"
)

# ============================================================
# GIT CONFIGURATION
# ============================================================

subprocess.run(
    ["git", "config", "--global", "user.name", USERNAME],
    check=False
)

subprocess.run(
    ["git", "config", "--global", "user.email", EMAIL],
    check=False
)

# Add or update origin
remotes = subprocess.run(
    ["git", "remote"],
    capture_output=True,
    text=True
).stdout.split()

if "origin" in remotes:
    subprocess.run(
        ["git", "remote", "set-url", "origin", REMOTE_URL],
        check=False
    )
else:
    subprocess.run(
        ["git", "remote", "add", "origin", REMOTE_URL],
        check=False
    )

print("✓ Git configured")

# ============================================================
# PROJECT STATUS
# ============================================================

print("\n==============================")
print("PROJECT STATUS")
print("==============================\n")

!git status

print("\n==============================")
print("CHANGED FILES")
print("==============================\n")

!git diff --stat

# ============================================================
# OPTIONAL BACKUP BRANCH
# ============================================================

if MAKE_BACKUP_BRANCH:

    backup_name = (
        "backup-"
        + datetime.now().strftime("%Y-%m-%d-%H%M")
    )

    print(f"\nCreating backup branch: {backup_name}")

    !git checkout -b {backup_name}
    !git push -u origin {backup_name}
    !git checkout main

    print("✓ Backup completed\n")

# ============================================================
# COMMIT MESSAGE
# ============================================================

print()

commit_message = input(
    "Commit message: "
).strip()

if commit_message == "":
    commit_message = f"Update {datetime.now():%Y-%m-%d %H:%M}"

# ============================================================
# COMMIT
# ============================================================

print("\nFiles that will be committed:\n")

!git status --short

!git add .

result = subprocess.run(
    [
        "git",
        "commit",
        "-m",
        commit_message
    ]
)

if result.returncode != 0:
    print("\nNothing to commit.")

# ============================================================
# PUSH
# ============================================================

print("\n==============================")
print("PUSH TO GITHUB")
print("==============================\n")

!git push origin main

print("\n==============================")
print("DONE")
print("==============================")
```
