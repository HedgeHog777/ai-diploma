```python
# ------------------------------------------------------------
# SINGLE-SPLIT RESULTS
# ------------------------------------------------------------

single_split = ml_results.copy()

single_split["Evaluation"] = "Single train/test split"
```

```python
# ------------------------------------------------------------
# WALK-FORWARD RESULTS
# ------------------------------------------------------------

walk_forward = validation_summary[
    ["model", "MAE_mean", "RMSE_mean", "R2_mean"]
].copy()

walk_forward = walk_forward.rename(columns={
    "model": "Model",
    "MAE_mean": "MAE",
    "RMSE_mean": "RMSE",
    "R2_mean": "R2"
})

walk_forward["Evaluation"] = "Walk-forward validation"

single_split_final = single_split[
    ["Model", "MAE", "RMSE", "R2", "Evaluation"]
].copy()
```

```python
# ------------------------------------------------------------
# COMBINED TABLE
# ------------------------------------------------------------

final_comparison = pd.concat(
    [
        single_split_final,
        walk_forward[
            ["Model", "MAE", "RMSE", "R2", "Evaluation"]
        ]
    ],
    ignore_index=True
)

display(
    final_comparison
    .sort_values(
        ["Evaluation", "R2"],
        ascending=[True, False]
    )
    .round(4)
)
```

```python
# ============================================================
# 009 — FINAL CANDIDATE MODELS
# ============================================================

print("=" * 70)
print("FINAL CANDIDATE MODELS")
print("=" * 70)

candidate_names = [
    "Random Forest — All Features",
    "7-Day Mean",
    "User Mean",
    "Linear Regression — Base",
    "Decision Tree — Base"
]

candidate_summary = validation_summary[
    validation_summary["model"].isin(candidate_names)
].copy()

candidate_summary = candidate_summary.sort_values(
    "R2_mean",
    ascending=False
)

display(
    candidate_summary.round(4)
)
```
