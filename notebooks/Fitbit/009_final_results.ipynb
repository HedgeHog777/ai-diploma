```python
# ============================================================
# 009 — FINAL RESULTS
# ============================================================
from pathlib import Path
import json
import pandas as pd
import numpy as np

from google.colab import drive
drive.mount("/content/drive")

print("=" * 70)
print("009 — FINAL RESULTS")
print("=" * 70)

PROJECT_ROOT = Path("/content/drive/MyDrive/FitnessML_Master")
TABLES_DIR = PROJECT_ROOT / "tables"
RESULTS_DIR = PROJECT_ROOT / "results"
OUTPUT_DIR = RESULTS_DIR / "009_final_results"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


def load_table(filename, required=True):
    path = TABLES_DIR / filename
    if not path.exists():
        if required:
            raise FileNotFoundError(f"Required table not found: {path}")
        return None
    return pd.read_csv(path)


def require_columns(df, columns, name):
    missing = set(columns) - set(df.columns)
    if missing:
        raise ValueError(
            f"{name}: missing columns {sorted(missing)}"
        )
```

```python
# ============================================================
# 009 — FINAL MODEL COMPARISON
# ============================================================

print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)
```

```python
# ------------------------------------------------------------
# SINGLE-SPLIT RESULTS
# ------------------------------------------------------------

single_split = (
    ml_results[
        ["Model", "MAE", "RMSE", "R2"]
    ]
    .copy()
)

single_split["Evaluation"] = "Single train/test split"
```

```python
# ------------------------------------------------------------
# WALK-FORWARD RESULTS
# ------------------------------------------------------------

walk_forward = validation_summary[
    ["model", "MAE_mean", "RMSE_mean", "R2_mean"]
].copy()

walk_forward = walk_forward.rename(columns={
    "model": "Model",
    "MAE_mean": "MAE",
    "RMSE_mean": "RMSE",
    "R2_mean": "R2"
})

walk_forward["Evaluation"] = "Walk-forward validation"

single_split_final = single_split[
    ["Model", "MAE", "RMSE", "R2", "Evaluation"]
].copy()
```

```python
# ------------------------------------------------------------
# COMBINED TABLE
# ------------------------------------------------------------

final_comparison = pd.concat(
    [
        single_split_final,
        walk_forward[
            ["Model", "MAE", "RMSE", "R2", "Evaluation"]
        ]
    ],
    ignore_index=True
)

display(
    final_comparison
    .sort_values(
        ["Evaluation", "R2"],
        ascending=[True, False]
    )
    .round(4)
)
```

```python
# ============================================================
# 009 — FINAL CANDIDATE MODELS
# ============================================================

print("=" * 70)
print("FINAL CANDIDATE MODELS")
print("=" * 70)

candidate_names = [
    "Random Forest — All Features",
    "7-Day Mean",
    "User Mean",
    "Linear Regression — Base",
    "Decision Tree — Base"
]

candidate_summary = validation_summary[
    validation_summary["model"].isin(candidate_names)
].copy()

candidate_summary = candidate_summary.sort_values(
    "R2_mean",
    ascending=False
)

display(
    candidate_summary.round(4)
)
```

```python
# ============================================================
# SAVE FINAL MODEL COMPARISON TABLES
# ============================================================

print("\n" + "=" * 70)
print("SAVING FINAL COMPARISON TABLES")
print("=" * 70)

# Use the existing project results directory
FINAL_DIR = RESULTS_DIR / "009_final_results"
FINAL_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# 1. Complete model comparison

final_comparison.to_csv(
    FINAL_DIR / "final_model_comparison.csv",
    index=False
)

print("✓ final_model_comparison.csv")
```

```python
# ------------------------------------------------------------
# 2. Final candidate models
#    = 006 walk-forward validation summary
# ------------------------------------------------------------

final_candidate_models = (
    validation_summary
    .copy()
    .sort_values("R2_mean", ascending=False)
    .reset_index(drop=True)
)

final_candidate_models.to_csv(
    FINAL_DIR / "final_candidate_models.csv",
    index=False
)

print("✓ final_candidate_models.csv")
```

```python
# ============================================================
# 009 — EXPORT ARTIFACTS FOR 010 UI
# ============================================================

print("\n" + "=" * 70)
print("EXPORTING ARTIFACTS FOR 010 UI")
print("=" * 70)

import shutil
import json

UI_SOURCE_DIR = RESULTS_DIR / "008_interpretability"
```

```python
# ------------------------------------------------------------
# 1. Final Random Forest model
# ------------------------------------------------------------

RF_MODEL_SOURCE = (
    UI_SOURCE_DIR / "random_forest_reference.pkl"
)

RF_MODEL_DEST = (
    FINAL_DIR / "random_forest_final.pkl"
)

if not RF_MODEL_SOURCE.exists():
    raise FileNotFoundError(
        f"Random Forest model not found:\n{RF_MODEL_SOURCE}"
    )

shutil.copy2(
    RF_MODEL_SOURCE,
    RF_MODEL_DEST
)

print("✓ random_forest_final.pkl")
```

```python
# ------------------------------------------------------------
# 2. Prediction results
# ------------------------------------------------------------

PREDICTIONS_SOURCE = (
    UI_SOURCE_DIR / "prediction_results.csv"
)

PREDICTIONS_DEST = (
    FINAL_DIR / "prediction_results.csv"
)

if not PREDICTIONS_SOURCE.exists():
    raise FileNotFoundError(
        f"Prediction results not found:\n{PREDICTIONS_SOURCE}"
    )

shutil.copy2(
    PREDICTIONS_SOURCE,
    PREDICTIONS_DEST
)

print("✓ prediction_results.csv")
```

```python
# ------------------------------------------------------------
# 3. Modeling dataset
# ------------------------------------------------------------

MODELING_SOURCE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "fitbit_modeling_features.csv"
)

MODELING_DEST = (
    FINAL_DIR / "fitbit_modeling_features.csv"
)

if not MODELING_SOURCE.exists():
    raise FileNotFoundError(
        f"Modeling dataset not found:\n{MODELING_SOURCE}"
    )

shutil.copy2(
    MODELING_SOURCE,
    MODELING_DEST
)

print("✓ fitbit_modeling_features.csv")
```

```python
# ------------------------------------------------------------
# 4. Model metadata
# ------------------------------------------------------------

METADATA_SOURCE = (
    UI_SOURCE_DIR / "model_metadata.json"
)

METADATA_DEST = (
    FINAL_DIR / "model_metadata.json"
)

if not METADATA_SOURCE.exists():
    raise FileNotFoundError(
        f"Model metadata not found:\n{METADATA_SOURCE}"
    )

with open(
    METADATA_SOURCE,
    "r",
    encoding="utf-8"
) as f:
    model_metadata = json.load(f)

model_metadata.update({
    "source_notebook": "009_final_results",
    "ui_model_file": "random_forest_final.pkl",
    "ui_prediction_file": "prediction_results.csv",
    "ui_modeling_data_file": "fitbit_modeling_features.csv"
})

with open(
    METADATA_DEST,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        model_metadata,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ model_metadata.json")
```

```python
# ------------------------------------------------------------
# 5. UI configuration
# ------------------------------------------------------------

ui_config = {
    "model": "Random Forest — All Features",
    "model_file": "random_forest_final.pkl",
    "prediction_file": "prediction_results.csv",
    "modeling_data_file": "fitbit_modeling_features.csv",
    "metadata_file": "model_metadata.json",
    "target": model_metadata.get(
        "target",
        "Calories"
    )
}

with open(
    FINAL_DIR / "ui_config.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        ui_config,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ ui_config.json")
```

```python
# ------------------------------------------------------------
# FINAL UI EXPORT CHECK
# ------------------------------------------------------------

print("\nUI artifacts:")

for path in [
    RF_MODEL_DEST,
    PREDICTIONS_DEST,
    MODELING_DEST,
    METADATA_DEST,
    FINAL_DIR / "ui_config.json"
]:
    print(f"  ✓ {path.name}")

print("=" * 70)
print("UI EXPORT COMPLETE")
print("=" * 70)
```
