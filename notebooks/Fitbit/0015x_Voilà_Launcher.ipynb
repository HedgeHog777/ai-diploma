```python
# @title
# ============================================================
# 0015x — VOILÀ LAUNCHER
# ============================================================

from google.colab import drive, output

import os
import subprocess
import sys
import time


# ------------------------------------------------------------
# GOOGLE DRIVE
# ------------------------------------------------------------

drive.mount(
    "/content/drive",
    force_remount=False
)


# ------------------------------------------------------------
# VOILÀ NOTEBOOK
# ------------------------------------------------------------

VOILA_NOTEBOOK = (
    "/content/drive/MyDrive/"
    "FitnessML_Master/"
    "notebooks/"
    "Fitbit/"
    "0014_VOILA.ipynb"
)

if not os.path.exists(VOILA_NOTEBOOK):
    raise FileNotFoundError(
        "Не знайдено 0014:\n"
        + VOILA_NOTEBOOK
    )
```

```python
# @title
# ------------------------------------------------------------
# ISOLATED VOILÀ ENVIRONMENT
# ------------------------------------------------------------

print()
print("=" * 60)
print("CREATING ISOLATED VOILÀ ENVIRONMENT")
print("=" * 60)

VOILA_ENV = "/content/voila_env"
VOILA_PYTHON = os.path.join(
    VOILA_ENV,
    "bin",
    "python"
)

# ------------------------------------------------------------
# CREATE VIRTUAL ENVIRONMENT
# ------------------------------------------------------------

if not os.path.exists(VOILA_PYTHON):

    print("Створення окремого Python environment...")

    result = subprocess.run(
        [
            sys.executable,
            "-m",
            "venv",
            "--system-site-packages",
            VOILA_ENV,
        ],
        capture_output=True,
        text=True,
    )

    print("return code:", result.returncode)

    if result.stdout:
        print("\n--- STDOUT ---")
        print(result.stdout)

    if result.stderr:
        print("\n--- STDERR ---")
        print(result.stderr)

    if result.returncode != 0:
        raise RuntimeError(
            "Не вдалося створити ізольоване Python environment."
        )

else:

    print("Окреме середовище вже існує.")

# ------------------------------------------------------------
# INSTALL ONLY INSIDE VENV
# ------------------------------------------------------------

print()
print("Встановлення Voilà у ізольоване середовище...")

install_result = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "--python",
        VOILA_PYTHON,
        "install",
        "--upgrade",
        "voila==0.5.13",
"ipywidgets==8.1.7",
"jupyterlab_widgets==3.0.15",
"widgetsnbextension==4.0.14",
        "widgetsnbextension==4.0.14",
"jupyterlab",
"notebook",
"ipykernel",
    ],
    capture_output=True,
    text=True,
)

print("return code:", install_result.returncode)

if install_result.stdout:
    print("\n--- PIP STDOUT ---")
    print(install_result.stdout)

if install_result.stderr:
    print("\n--- PIP STDERR ---")
    print(install_result.stderr)

if install_result.returncode != 0:
    raise RuntimeError(
        "Не вдалося встановити Voilà у ізольоване середовище."
    )

# ------------------------------------------------------------
# VERIFY VENV
# ------------------------------------------------------------

print()
print("=" * 60)
print("ISOLATED VOILÀ ENVIRONMENT")
print("=" * 60)

subprocess.run(
    [
        VOILA_PYTHON,
        "-c",
        """
import sys
import voila
import ipywidgets
import jupyterlab_widgets

print("Python:", sys.executable)
print("Voilà:", voila.__version__)
print("ipywidgets:", ipywidgets.__version__)
print("jupyterlab_widgets:", jupyterlab_widgets.__version__)
""",
    ],
    check=True,
)

print("=" * 60)
```

```python
# @title
# ------------------------------------------------------------
# VERIFY WIDGET RUNTIME
# ------------------------------------------------------------

widget_check = subprocess.run(
    [
        VOILA_PYTHON,
        "-c",
        """
import ipywidgets
import jupyterlab_widgets
import voila

print("ipywidgets:", ipywidgets.__version__)
print("jupyterlab_widgets:", jupyterlab_widgets.__version__)
print("voila:", voila.__version__)

w = ipywidgets.HTML(value="<b>Widget test OK</b>")
print("widget model:", w._model_name)
print("widget view:", w._view_name)
""",
    ],
    capture_output=True,
    text=True,
)

print(widget_check.stdout)
print(widget_check.stderr)

if widget_check.returncode != 0:
    raise RuntimeError("Widget runtime test failed.")

print("✓ Widget runtime OK")
```

```python
# @title
# ------------------------------------------------------------
# START VOILÀ
# ------------------------------------------------------------

PORT = 8866

process = subprocess.Popen(
    [
     VOILA_PYTHON,
        "-m",
        "voila",
        VOILA_NOTEBOOK,
        "--port",
        str(PORT),
        "--no-browser",
 "--Voila.ip=0.0.0.0",
        "--strip_sources=True",
      "--theme=dark",
        "--show_tracebacks=True",
        "--VoilaExecutor.startup_timeout=120",

#"--ServerApp.allow_origin=*",
#"--ServerApp.allow_websocket_origin=*",
#"--ServerApp.allow_remote_access=True",
#"--ServerApp.disable_check_xsrf=True",


    ],

    stdout=log_file,
    stderr=subprocess.STDOUT,
    env=voila_env,
)


print()
print(
    f"Voilà process started. PID: {process.pid}"
)


# ------------------------------------------------------------
# WAIT FOR SERVER
# ------------------------------------------------------------

time.sleep(7)

print(
    subprocess.run(
        [
            "curl",
            "-I",
            f"http://127.0.0.1:{PORT}"
        ],
        capture_output=True,
        text=True
    ).stdout
)
# ------------------------------------------------------------
# CHECK PROCESS
# ------------------------------------------------------------

if process.poll() is not None:

    log_file.flush()

    with open(
        LOG_PATH,
        "r",
        encoding="utf-8",
        errors="replace"
    ) as f:

        log_text = f.read()

    raise RuntimeError(
        "Voilà завершив роботу.\n\n"
        + log_text[-12000:]
    )
```
