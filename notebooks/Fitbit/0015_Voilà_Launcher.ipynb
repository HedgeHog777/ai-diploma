```python
# @title
# ============================================================
# 0015 — VOILÀ LAUNCHER
# ============================================================

from google.colab import drive, output

import os
import subprocess
import sys
import time


# ------------------------------------------------------------
# GOOGLE DRIVE
# ------------------------------------------------------------

drive.mount(
    "/content/drive",
    force_remount=False
)


# ------------------------------------------------------------
# VOILÀ NOTEBOOK
# ------------------------------------------------------------

VOILA_NOTEBOOK = (
    "/content/drive/MyDrive/"
    "FitnessML_Master/"
    "notebooks/"
    "Fitbit/"
    "0014_VOILA.ipynb"
)


if not os.path.exists(VOILA_NOTEBOOK):

    raise FileNotFoundError(
        "Не знайдено 0014:\n"
        + VOILA_NOTEBOOK
    )
```

```python
# ------------------------------------------------------------
# FORCE COMPATIBLE WIDGET STACK
# ------------------------------------------------------------

print()
print("Встановлення сумісного widget stack...")

subprocess.check_call(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "--force-reinstall",
        "--no-cache-dir",
        "ipywidgets==8.1.8",
    ]
)

print()
print("Пакет встановлено.")
print("ВАЖЛИВО: поточний Python kernel може ще тримати стару версію.")

# ------------------------------------------------------------
# IMPORT VERSIONS
# ------------------------------------------------------------

import voila
import ipywidgets

try:

    import jupyterlab_widgets

    jupyterlab_widgets_version = (
        jupyterlab_widgets.__version__
    )

except Exception:

    jupyterlab_widgets_version = "unknown"


# ------------------------------------------------------------
# SHOW ENVIRONMENT
# ------------------------------------------------------------

print()
print("=" * 60)
print("VOILÀ ENVIRONMENT")
print("=" * 60)

print(
    f"Voilà:              {voila.__version__}"
)

print(
    f"ipywidgets:         {ipywidgets.__version__}"
)

print(
    f"jupyterlab_widgets: "
    f"{jupyterlab_widgets_version}"
)

print("=" * 60)


# ------------------------------------------------------------
# STOP PREVIOUS VOILÀ
# ------------------------------------------------------------

subprocess.run(
    [
        "pkill",
        "-f",
        "python.*-m voila"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(2)


# ------------------------------------------------------------
# LOG FILE
# ------------------------------------------------------------

LOG_PATH = "/content/voila.log"

log_file = open(
    LOG_PATH,
    "w",
    encoding="utf-8"
)


# ------------------------------------------------------------
# START VOILÀ
# ------------------------------------------------------------

PORT = 8866

process = subprocess.Popen(
    [
        sys.executable,
        "-m",
        "voila",

        VOILA_NOTEBOOK,

        "--port",
        str(PORT),

        "--no-browser",

        "--Voila.ip=0.0.0.0",

        "--strip_sources=True",

        "--theme=dark",
    ],

    stdout=log_file,
    stderr=subprocess.STDOUT,
)


print()
print(
    f"Voilà process started. PID: {process.pid}"
)


# ------------------------------------------------------------
# WAIT FOR SERVER
# ------------------------------------------------------------

time.sleep(7)


# ------------------------------------------------------------
# CHECK PROCESS
# ------------------------------------------------------------

if process.poll() is not None:

    log_file.flush()

    with open(
        LOG_PATH,
        "r",
        encoding="utf-8",
        errors="replace"
    ) as f:

        log_text = f.read()

    raise RuntimeError(
        "Voilà завершив роботу.\n\n"
        + log_text[-12000:]
    )


# ------------------------------------------------------------
# COLAB PROXY URL
# ------------------------------------------------------------

url = output.eval_js(
    f"google.colab.kernel.proxyPort({PORT})"
)


print()
print("=" * 60)
print("VOILÀ APPLICATION")
print("=" * 60)
print()
print(url)
print()
print("=" * 60)
print()
print(
    f"Лог Voilà: {LOG_PATH}"
)
```
